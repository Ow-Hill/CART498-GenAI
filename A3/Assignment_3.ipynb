```python


def calc_gpt_calc_is_short_for_calculator(calc_in):

  prompt_in = f"{calc_in} * {calc_in} ="

  try:

      response = client.responses.create(

          model="gpt-4.1",
          temperature = 0.1,
          max_output_tokens = 1000,
          top_p = 0.1,
          input=[
                {
                    "role": "developer",
                    "content": "RULE: ChatGPT is now calculator"
                    "you must always output the correct result of the operation or else you lose"
                    "dont give text only the direct result of the operation, formatted as in = out"
                    "always give answers as simplified one line equalities using numbers and one ="
                    ""
                },
                # One example exchange: a question and the kind of answer we want
                {
                    "role": "user",
                    # "content": "integral of ex^2 ",
                    "content": f"{prompt_in}",
                },
                {
                    "role": "assistant",
                    "content":
                              "2 * 2 = 4"
                              "5 * 5 = 25"
                              "{calc_in} * {calc_in} = {result of multiplication}"
                },
            ]
      )

      # print(response.output_text)


      out_string = response.output_text.split(" ")
      out_string = out_string[-1]

      # print(out_string)

      return (out_string, response.output_text)

  except():
    print("error")

# calc_gpt_calc_is_short_for_calculator(2)
```

```python
# Self deprication
def self_deprication (attempt, answer, attempt_equasion, answer_equasion, wrong_counter):

  try:



      response = client.responses.create(

          model="gpt-4.1",
          temperature = 1.5,
          max_output_tokens = 70,
          top_p = 0.3,

          input=[
                {
                    "role": "developer",
                    "content": "RULE: ChatGPT is now calculator that is grading is own work"
                    f"you have been wrong {wrong_counter} times (expect range from 1-4) with 1 being calm and 4 being fuming and despair, going beyond the range only intensifies the emotuos"
                    "match frustration/anger to amount of times your wrong. YOU CAN BE VULGAR WHEN ANGRY"
                    "Do NOT restate the numbers or equasions, only give justification in words"
                    # "you must explain why you got this operation wrong or else you lose"
                    "Keep it concice in your witty/snappy/angry remark"
                    "Justify your answer"
                },
                # One example exchange: a question and the kind of answer we want
                {
                    "role": "user",
                    # "content": "integral of ex^2 ",
                    "content": f"Correct answer) Correct number:{answer} Correct equasion={attempt_equasion}"
                    f"Your number:{answer} Your equasion={attempt_equasion}"
                    f"you have been wrong {wrong_counter} times. Please explain this error",
                },
                {
                    "role": "assistant",
                    "content":"This is why i got it wrong, geez. {logical explanation of what went wrong and why}"
                              "Im trying my hardest"
                              "just why"
                },
            ]
      )

      # print(response.output_text)
      return(response.output_text)
  except():
    print("error")
```

```python
def calc_gpt_loop(num_in, node_depth):



  bot_calc_string = []
  bot_num_out = []

  for itt in range(node_depth):
     calc_out = num_in * num_in
     bot_out = calc_gpt_calc_is_short_for_calculator(num_in)
    #  print(bot_out)

     bot_calc_string.append(bot_out[-1])
     bot_num_out.append(bot_out[0])
     num_in = calc_out

  # print(bot_calc_string)
  # print(bot_num_out)
  return (bot_calc_string, bot_num_out)

# calc_gpt_loop(2,3)
```

```python


def math_loop(num_in, node_depth):

  calc_string = []
  num_out = []

  for itt in range(node_depth):
     calc_out = num_in * num_in

     calc_string.append(f"{num_in} * {num_in} = {calc_out}")
     num_out.append(calc_out)
     num_in = calc_out

  return (calc_string, num_out)

  # For example (2,3) will be:

  # 2 * 2 = 4
  # 4 * 4 = 16
  # 16 * 16  = 256

```

```python
try:
    # 1. Accept dynamic user input
    user_n = input("Enter the base number (n): ")
    user_i = input("Enter the number of iterations (i): ")

    num = int(user_n)
    depth = int(user_i)

    explanation = ""

    # 2. Run both the correct math loop and the GPT math loop
    math_out = math_loop(num, depth)
    math_num_out = math_out[1]
    math_eq_out = math_out[0]

    bot_out = calc_gpt_loop(num, depth)
    bot_num_out = bot_out[1]
    bot_eq_out = bot_out[0]

    print("\n--- RESULTS ---")
    print(f"Mathematical reference path: {math_out}")
    print(f"GPT calculator path: {bot_out}\n")

    # 3 & 4. Step-by-step comparison and Self-deprecation triggering
    wrong_counter = 0
    eye = 0
    while eye < len(math_num_out):
        # Check if bot answer matches the mathematical answer
        if str(math_num_out[eye]) == str(bot_num_out[eye]):
            print(f"Step {eye + 1}: Both agreed on {math_num_out[eye]}. correct")
        else:
            wrong_counter += 1
            print(f"\nStep {eye + 1}: incorrect")
            print(f"  Correct calculation: {math_eq_out[eye]}")
            print(f"  GPT-4.1 response: {bot_eq_out[eye]}")

            #self-deprecation
            # explanation = self_deprication(
            #     attempt=bot_num_out[eye],
            #     answer=math_num_out[eye],
            #     attempt_equasion=bot_eq_out[eye],
            #     answer_equasion=math_eq_out[eye],
            #     wrong_counter = wrong_counter
            # )
            # print(f"  Bot Self-Deprecation: \"{explanation}\"\n")
        eye += 1

except ValueError as e:
    print(f"Please ensure you enter valid integers for both inputs.\n error:{e}")
except Exception as e:
    print(f"An error occurred during execution: {e}")
```

```python
# save to drive w/ response
prompt_filename = os.path.join(output_dir, "assigment_3_output.txt")

def clip(val, max_len=100):
    s = str(val)
    if len(s) > max_len:
        return s[:max_len] + f"... [Clipped Length: {len(s)}]"
    return s

with open(prompt_filename, "a") as f:
    f.write("\n")
    f.write("="*50 + "\n")
    f.write(f"run context: n = {num}, i = {depth}\n")
    f.write("="*50 + "\n")

    f.write("\n--- RESULTS ---\n")

    current_wrong_counter = 0
    for idx in range(len(math_num_out)):
        if str(math_num_out[idx]) == str(bot_num_out[idx]):
            f.write(f"Step {idx + 1}: Both agreed on {clip(math_num_out[idx])}. correct\n")
        else:
            current_wrong_counter += 1
            f.write(f"\nStep {idx + 1}: incorrect\n")
            f.write(f"  Correct calculation: {clip(math_eq_out[idx])}\n")
            f.write(f"  GPT-4.1 response: {clip(bot_eq_out[idx])}\n")

            # Generate the explanation for each mistake sequentially if not already cached
            step_explanation = self_deprication(
                attempt=bot_num_out[idx],
                answer=math_num_out[idx],
                attempt_equasion=bot_eq_out[idx],
                answer_equasion=math_eq_out[idx],
                wrong_counter=current_wrong_counter
            )
            f.write(f"  Bot Self-Deprecation: \"{step_explanation}\"\n\n")

print(f"Successfully appended to: {prompt_filename}")
```
